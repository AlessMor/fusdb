# Interactive Atomic Physics Rate Plotter

AMJUEL rate coefficients for every atomic-physics relation in the registry —
H.2 fits $\langle \sigma v \rangle(T)$ and H.4 fits
$\langle \sigma v \rangle(n_e, T)$ — drawn as a client-side interactive
Bokeh explorer (`fusdb.plotting.atomic_physics`). The density slider moves
the `[H.4]` curves across the AMJUEL density range; filter by process and
species (hydrogenic species are shown by default, one click adds He or the
impurities), click legend entries to hide curves, and pan/zoom or set
explicit axis limits. Hovering a curve shows its reaction and, where AMJUEL
reports one, the fit error.

The last cell saves the standalone widget embedded by the documentation page
`docs/knowledge_base/plasma_physics/7-boundary_and_sheath_physics/electrons_ions_neutrals_interactions.md`.

```python
from pathlib import Path
import sys

from bokeh.io import output_notebook, show

root = Path.cwd().resolve()
while root != root.parent and not (root / "src" / "fusdb").is_dir():
    root = root.parent

src_path = str(root / "src")
if src_path not in sys.path:
    sys.path.insert(0, src_path)

from fusdb.plotting.atomic_physics import atomic_physics_app, save_atomic_physics_app_html

output_notebook()
```
