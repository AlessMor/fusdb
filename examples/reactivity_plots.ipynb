# Interactive Reactivity Plotter

Fusion reactivities $\langle \sigma v \rangle(T_i)$ for every reaction and
parametrisation registered in the relation registry, drawn as a client-side
interactive Bokeh explorer (`fusdb.plotting.reactivity`). Filter by reaction
and source, click legend entries to hide curves, and pan/zoom or set explicit
axis limits.

The last cell saves the standalone widget embedded by the documentation page
`docs/knowledge_base/plasma_physics/0-fusion_plasmas/cross_sections_reactivities.md`.

```python
from pathlib import Path
import sys

from bokeh.embed import file_html
from bokeh.io import output_notebook, show
from bokeh.resources import CDN

root = Path.cwd().resolve()
while root != root.parent and not (root / "src" / "fusdb").is_dir():
    root = root.parent

src_path = str(root / "src")
if src_path not in sys.path:
    sys.path.insert(0, src_path)

from fusdb.plotting.reactivity import reactivity_app

output_notebook()
```

```python
show(reactivity_app())
```

```python
out_path = root / "docs" / "code_docs" / "reactivity_plotter.html"
out_path.parent.mkdir(parents=True, exist_ok=True)
out_path.write_text(file_html(reactivity_app(num_points=400), CDN, "Fusion Reactivity Plotter"), encoding="utf-8")
print(f"Wrote: {out_path}")
```
